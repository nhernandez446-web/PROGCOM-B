```python
helado=input("Quieres helado?")
helaresta="Te lo compro"
helanegacion="No te lo compro"
match helado.lower():
  case "si":print(helaresta)
  case "sí":print(helaresta)
  case "sipis":print(helaresta)
  case "sisas":print(helaresta)
  case "clarinetes":print(helaresta)
  case "seh":print(helaresta)
  case _:print(helanegacion)
```
