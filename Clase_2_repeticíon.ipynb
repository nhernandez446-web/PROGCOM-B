# Estructuras de repetición
Es utilizada para repetecir una opción n cantidad de veces o hasta que se logre un acondición.

##Estructura PARA (FOR)
Se usa para repetir una acción cierta cantidad de veces.

for (variable) in range (Valor Inicial, Valor Final+1,Paso): Se repite (VI-VF)/Pasos

```python
#Programa que imprima los múltiplos de 3, que se encuentren del 1 al 30

for inter in range(1,31):
  if inter%3==0:
    print(inter)
```

```python
for interp in range (3,31,3):
  print(interp)
```

```python
#Despegue
for desp in range(10,0,-1):
  print(desp)
```

```python
for cua in range(1,11):
  print(f"4 * {cua} = {cua*4}") #< mejor que hacer la operación aparte.
```

```python
#tabla de n

nume=int(input("Por favor inserte un número: "))
for i in range(0,10+1):
  print(f"{nume} * {i} = {nume*i}")
```

```python
for num in range(1,11):
  print(f"Tabla del {num}")
  for i in range(1,11):
    print(f"{num} * {i} = {num*i}")
```

```python
n=int(input("Por favor inserte un número "))
print(f"Tabla del {n}")
c=0
for em in range(n,11*n,n):
  c+=1
  print(f"{n} * {c} = {em}")
```

##Estructura mientras (WHILE)
Estructura computacional de repetición, empleada para repetir una acci+on **MIENTRAS** se cumpla una condición.

while (condición):
  * Ejecuta el código
  * Modificador de la condición

While acondicional:
---
while True:
  * Ejecuta el código
  * Modificador de condición
  * break()

```python
#Programa que imprime los números pares
num=2
while num<=20:
  print(num)
  num+=2 #contador del para
```

```python
#Tabla de multiplicar del 5
print("Tabla del 5")
i=1
while i<=10:
  print(f"5 * {i} = {i*5}")
  i+=1 #contador
```
