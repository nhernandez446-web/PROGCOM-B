```python
# Mascotas virtual
class Mascota():
  def __init__(self,nombre,raza,energia):
    self.nombre=nombre
    self.raza=raza
    self.hambre = 50
    self.energia = energia
    self.vivo = True

  def saludar(self):
    print(f"Hola, soy {self.nombre}")

  def alimentar(self,c):
    comida={"leche":10,"polvo":50}
    self.hambre -= comida[c]
    self.energia += comida[c]/2
    print(f"{self.nombre} comio {c}")

  def jugar(self):
    if self.energia <= 0:
      print(f"{self.nombre} no tiene energia, debe dormir")
      return
    self.energia -= 10
    print(f"{self.nombre} ha jugado. Energia restante: {self.energia}")

  def dormir(self):
    if self.energia <= 0:
      print(f"{self.nombre} esta agotado y se va a dormir 😴")
    self.energia = 100
    print(f"{self.nombre} desperto con energia!")

  def revivir(self):
    self.energia = 100
    self.vivo = True
    print(f"¡{self.nombre} ha revivido! ✨ Está como nuevo!")

mascota1 = Mascota("Juan", "perro", 200)
mascota2 = Mascota("Chefsito", "Rata", 50)
```
